```python
# AND
import numpy as np
def And(x1, x2):
    w1, w2, theta = 0.5, 0.5, 0.7
    if w1*x1 + w2*x2 >= theta:
        return 1
    else:
        return 0

def AND(x1, x2):
    x = np.array([x1, x2])
    w= np.array([0.5, 0.5])
    b = -0.7
    tmp = np.sum(x * w) + b
    if tmp > 0:
        return 1
    else:
        return 0
    
    
test_cases = [
    (0, 0),
    (0, 1),
    (1, 0),
    (1, 1)
]

print("x1 x2 | And AND")
for x1, x2 in test_cases:
    print(x1, x2, "|", And(x1, x2), AND(x1, x2))
```

```python
# NAND

def NAND(x1, x2):
    x = np.array([x1, x2])
    w = np.array([-0.5, -0.5])
    b = 0.7
    tmp = np.sum(x * w) + b
    if tmp > 0 :
        return 1
    else:
        return 0
    
test_cases = [
    (0, 0),
    (0, 1),
    (1, 0),
    (1, 1)
]

print("x1 x2 | NAND")
for (x1, x2) in test_cases:
    print(x1, x2, "|", NAND(x1, x2))
```

```python
# OR

def OR(x1, x2):
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = -0.3
    tmp = np.sum(w*x) + b
    if tmp <= 0:
        return 0
    else:
        return 1
    
test_cases = [
    (0, 0),
    (0, 1),
    (1, 0),
    (1, 1)
]

print("x1 x2 | OR")
for (x1, x2) in test_cases:
    print(x1, x2, "|", OR(x1, x2))
```

```python
# XOR

def XOR(x1, x2):
    s1 = NAND(x1, x2)
    s2 = OR(x1, x2)
    y = AND(s1, s2)
    return y

test_cases = [
    (0, 0),
    (0, 1),
    (1, 0),
    (1, 1)
]

print("x1 x2 | XOR")
for (x1, x2) in test_cases:
    print(x1, x2, "|", XOR(x1, x2))
```

```python
# step function

def step_function(x):
    y = x > 0
    return y.astype(np.int32)

import numpy as np
import matplotlib.pyplot as plt

fig, ax = plt.subplots()

x = np.linspace(-10,10,100)
y = step_function(x)

ax.plot(x, y)
```

```python
# ReLU

def ReLU (x):
    return np.maximum(x, 0)

fig, ax = plt.subplots()
x = np.linspace(-10,10,100)
y = ReLU(x)
ax.plot(x, y)
```

```python
#sigmoid

import numpy as np
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
x_val = np.linspace(-10,10,100)
y_val = 1 / (1 + np.exp(-1 * x_val))
ax.plot(x_val, y_val )
```

```python
# softmax

def softmax(a):
    c = np.max(a)
    a_exp = np.exp(a - c)
    sum_exp = np.sum(a_exp)
    return a_exp / sum_exp

a = np.array([1000, 1001, 1002])

y = softmax(a)

print(y)
print(np.sum(y)) # 합은 1이 되어야함.
```
